### 1. What is Statistical Machine Translation (SMT)?

**Statistical Machine Translation (SMT)** is a machine translation paradigm that generates translations based on statistical models whose parameters are derived from the analysis of bilingual text corpora. Instead of relying on hand-crafted linguistic rules, SMT treats translation as a machine learning problem using probability distributions—specifically combining a **Translation Model** $P(f|e)$ (measuring how likely a source phrase corresponds to a target phrase) and a **Language Model** $P(e)$ (measuring the fluency of the target sentence) via Bayes' theorem: $\arg\max_e P(e|f) = \arg\max_e P(f|e)P(e)$.

### 2. What are the main differences between SMT and Neural Machine Translation (NMT)?

The primary differences between SMT and NMT include:

* **Architecture:** SMT uses separate, independently trained components (alignment models, phrase tables, language models, reordering models), whereas NMT uses a single, end-to-end neural network (typically an Encoder-Decoder or Transformer).
* **Representation:** SMT relies on discrete symbolic representations (words and phrases), while NMT maps words into continuous vector spaces (embeddings), capturing semantic similarity.
* **Context Handling:** SMT translates local phrases with limited context windows, often struggling with long-range dependencies. NMT captures whole-sentence context using recurrent layers or self-attention.
* **Fluency:** NMT generally produces much smoother, more natural translations than SMT, though SMT can sometimes be more predictable on rare domain-specific terms.

### 3. Explain the concept of attention in Neural Machine Translation.

**Attention** in NMT is a mechanism that allows the decoder to dynamically focus on different, relevant parts of the source sentence at each step of generating the target translation. Instead of compressing the entire source sentence into a single fixed-length context vector (which causes an information bottleneck), attention computes alignment weights over all encoder hidden states, creating a tailored weighted-sum context vector for every target word generated.

### 4. How do Generative Pre-trained Transformers (GPTs) contribute to machine translation?

**Generative Pre-trained Transformers (GPTs)** contribute to machine translation through massive self-supervised pre-training on multilingual text corpora, enabling:

* **Zero-shot and Few-shot Translation:** GPTs can translate between language pairs without task-specific fine-tuning simply by prompting (e.g., *"Translate English to French: Hello ->"*).
* **Contextual & Nuanced Translation:** With deep self-attention and large context windows, GPTs maintain document-level coherence, tone, and idiomatic accuracy better than traditional sentence-by-sentence models.
* **Multilingual Transfer:** Shared subword vocabularies and representations allow knowledge transfer across high-resource and low-resource languages.

### 5. What is poetry generation in generative AI?

**Poetry generation** in generative AI is the automated creation of poetic text that adheres to stylistic, structural, thematic, and rhythmic constraints (such as meter, rhyme schemes, stanza structure, imagery, and emotional tone). Models like GPT-2/GPT-4 or fine-tuned sequence models learn patterns of poetic language, metaphor, and phonetics from literary corpora to compose original verses from prompts or themes.

### 6. How does music composition with generative AI work?

**Music composition with generative AI** works by modeling musical sequences—either symbolically (MIDI events, pitch, duration, velocity, harmony) or directly as raw audio waveforms/spectrograms:

1. **Tokenization/Representation:** Musical notes, chords, and timing are converted into discrete tokens (similar to words in NLP) or continuous latent representations.
2. **Sequence/Diffusion Modeling:** Architectures such as Transformers (e.g., Music Transformer), RNNs/LSTMs, VAEs, or Diffusion models predict the next musical event or denoise audio conditioned on melody, genre, tempo, or text prompts.
3. **Synthesis/Rendering:** Symbolic outputs (MIDI) are rendered via virtual instruments, while audio models decode spectrograms back into playable waveforms.

### 7. What role does reinforcement learning play in generative AI for NLP?

**Reinforcement Learning (RL)** aligns generative NLP models with human preferences and non-differentiable sequence-level objectives:

* **Direct Metric Optimization:** Standard cross-entropy loss trains token-by-token, whereas RL (e.g., REINFORCE, PPO) directly optimizes sequence-level metrics like BLEU, ROUGE, or custom reward scores.
* **RLHF (Reinforcement Learning from Human Feedback):** Models are fine-tuned using a reward model trained on human rankings to make outputs more helpful, truthful, safe, and coherent.
* **Mitigating Exposure Bias:** RL allows the model to explore its own generated trajectories during training rather than relying solely on ground-truth teacher forcing.

### 8. What are multimodal generative models?

**Multimodal generative models** are AI systems capable of processing, understanding, and generating content across multiple data modalities—such as text, images, audio, and video—within a unified architecture. By projecting different modalities into a shared latent embedding space, these models perform cross-modal tasks like image captioning (image-to-text), text-to-image synthesis, visual question answering, and audio-visual generation.

### 9. Define Natural Language Understanding (NLU) in the context of generative AI.

In the context of generative AI, **Natural Language Understanding (NLU)** refers to the model's ability to accurately comprehend the semantic meaning, syntactic structure, user intent, entity relationships, sentiment, and implicit context of an input prompt so that it can generate coherent, contextually appropriate, and logically consistent responses.

### 10. What ethical considerations arise in generative AI for creative writing?

Key ethical considerations in generative AI for creative writing include:

* **Copyright and Intellectual Property:** Training on copyrighted books and articles without explicit consent raises legal and ethical questions regarding derivative works and fair compensation for authors.
* **Plagiarism and Attribution:** Models may inadvertently memorize and reproduce verbatim passages or closely mimic a living author's distinctive style.
* **Authenticity and Deception:** Passing AI-generated stories, articles, or essays as human-authored work undermines trust in creative industries and journalism.
* **Bias and Harm:** Generated narratives can perpetuate stereotypes, cultural misrepresentations, or harmful tropes present in training data.

### 11. How can attention mechanisms improve NMT performance on longer sentences?

Traditional sequence-to-sequence models encode an entire input sentence into a single fixed-size vector, causing early information to degrade as sentence length increases. **Attention mechanisms** overcome this bottleneck by:

* Retaining all encoder hidden states across every time step.
* Computing direct relevance scores between the current decoder state and every source token regardless of distance.
* Creating dynamic shortcuts for gradient flow during backpropagation and direct access to distant words during inference, preventing information loss in long sentences.

### 12. What are some challenges with bias in generative AI for machine translation?

Challenges with bias in generative AI for machine translation include:

* **Gender Bias:** Translating gender-neutral terms (e.g., *"the doctor"* or *"the nurse"*) into gendered languages (like Spanish, French, or German) often defaults to stereotypical gender roles (male doctors, female nurses).
* **Cultural and Dialectal Bias:** Models favor dominant dialects (e.g., US English or Castilian Spanish) while misinterpreting or erasing regional dialects and cultural idioms.
* **Formality and Social Hierarchy Errors:** Misassigning formal vs. informal pronouns (e.g., *tu/vous* in French or honorifics in Japanese/Korean) based on biased demographic assumptions.

### 13. Explain how reinforcement learning differs from supervised learning in generative AI.

* **Training Signal:** Supervised learning relies on labeled input-output pairs and minimizes token-level cross-entropy loss against ground-truth targets. Reinforcement learning learns from scalar **reward signals** evaluating an entire generated sequence.
* **Exploration vs. Imitation:** Supervised learning strictly imitates the training dataset (teacher forcing), which can lead to exposure bias. RL explores diverse generated outputs and reinforces high-reward sequences even when there is no single "correct" answer.
* **Objective Alignment:** RL can optimize non-differentiable human preferences (helpfulness, safety, style) that cannot be captured by exact word matching.

### 14. What is the role of a decoder in NMT models?

The **decoder** in an NMT model is responsible for generating the translated sequence in the target language autoregressively (one token at a time). At each time step, it:

1. Takes the previously generated target tokens and its internal hidden state.
2. Uses attention (cross-attention in Transformers) to query the encoder's representations of the source sentence.
3. Outputs a probability distribution over the target vocabulary to predict the next most likely word until an end-of-sequence (`<EOS>`) token is produced.

### 15. How does fine-tuning a GPT model differ from pre-training it?

* **Pre-training:** Conducted on massive, diverse, unlabeled text corpora using self-supervised learning (next-token prediction). It teaches the model general grammar, world knowledge, and reasoning at high computational cost.
* **Fine-tuning:** Conducted on a smaller, curated, task-specific or domain-specific dataset (e.g., parallel translation pairs or instruction-response pairs) with a lower learning rate. It adapts the pre-trained model's general capabilities to specialize in a target task, style, or format.

### 16. Describe one approach generative AI uses to avoid overfitting in creative content generation.

One widely used approach is **Dropout regularization combined with Stochastic Decoding (Top-k / Top-p Nucleus Sampling and Temperature scaling)**:

* During training, **Dropout** randomly deactivates a fraction of neural connections, preventing the model from memorizing exact training sequences.
* During inference, **Nucleus (Top-p) Sampling** samples from the smallest set of tokens whose cumulative probability exceeds a threshold $p$ (e.g., $p=0.9$), avoiding repetitive, deterministic memorization while keeping outputs coherent and creative.

### 17. What makes GPT-based models effective for creative storytelling?

GPT-based models excel at creative storytelling because of:

* **Large Context Windows & Self-Attention:** They track characters, plot threads, setting details, and narrative tone across long passages.
* **Vast World Knowledge & Style Diversity:** Pre-training on diverse literature enables them to adopt varied genres, voices, and narrative structures.
* **Autoregressive Coherence with Controlled Randomness:** Temperature and nucleus sampling allow them to balance logical plot progression with imaginative, unexpected phrasing.

### 18. How does context preservation work in NMT models?

Context preservation in NMT models works through:

* **Bidirectional Encoders (RNNs):** Process the source sentence forward and backward so each token representation encodes both preceding and succeeding words.
* **Self-Attention & Cross-Attention (Transformers):** Every token directly attends to every other token in the sequence, capturing long-distance syntactic and semantic dependencies.
* **Document-Level Context / Positional Encodings:** Positional embeddings preserve word order, while extended context windows allow models to reference surrounding sentences for pronoun resolution and consistent terminology.

### 19. What is the main advantage of multimodal models in creative applications?

The main advantage of multimodal models in creative applications is **seamless cross-modal translation and unified creative workflows**—allowing creators to use natural language, sketches, images, or audio interchangeably to ideate, generate, and edit rich media (e.g., generating illustrations for a story, scoring music to match a video scene, or creating interactive multimedia experiences) within a single coherent system.

### 20. How does generative AI handle cultural nuances in translation?

Generative AI handles cultural nuances by leveraging **contextual embeddings learned from vast multilingual and culturally diverse corpora**:

* Rather than translating idioms literally word-for-word, attention mechanisms recognize idiomatic expressions as whole semantic units and map them to culturally equivalent idioms in the target language.
* Through prompt conditioning or instruction tuning, models can adjust register, honorifics, regional dialects, and cultural references to suit the target audience.

### 21. Why is it difficult to fully remove bias in generative AI models?

Completely removing bias from generative AI models is difficult because:

* **Biased Training Data:** Human language and historical texts inherently reflect societal, cultural, and historical biases, which models learn as statistical patterns.
* **Subjectivity of Fairness:** Definitions of fairness and neutrality vary across cultures, languages, and contexts—eliminating one form of bias can inadvertently introduce another or degrade model utility.
* **Deep Entanglement in Representations:** Bias is not isolated to single neurons or words; it is distributed across high-dimensional semantic associations throughout the model's weights.

```python
# Practical 1: Implement a basic Statistical Machine Translation (SMT) model that uses word-by-word translation with a dictionary lookup approach.

from collections import defaultdict, Counter

class BasicSMTModel:
    def __init__(self):
        # Stores word translation probabilities P(target | source)
        self.translation_probs = {}
        # Unigram language model probabilities P(target)
        self.language_model = {}

    def train(self, parallel_corpus):
        """
        Trains a simple IBM Model 1-style word translation table and target language model
        from aligned (source, target) sentence pairs.
        """
        co_occurrence = defaultdict(Counter)
        source_counts = Counter()
        target_counts = Counter()
        total_target_words = 0

        for src_sent, tgt_sent in parallel_corpus:
            src_words = src_sent.lower().split()
            tgt_words = tgt_sent.lower().split()

            for tw in tgt_words:
                target_counts[tw] += 1
                total_target_words += 1

            for sw in src_words:
                source_counts[sw] += 1
                for tw in tgt_words:
                    co_occurrence[sw][tw] += 1

        # Compute P(target | source)
        for sw, tgt_counter in co_occurrence.items():
            self.translation_probs[sw] = {
                tw: count / source_counts[sw]
                for tw, count in tgt_counter.items()
            }

        # Compute target unigram probabilities P(target)
        self.language_model = {
            tw: count / total_target_words
            for tw, count in target_counts.items()
        }

    def translate(self, sentence):
        """Translates a sentence word-by-word using noisy-channel scoring: argmax P(t|s) * P(t)."""
        words = sentence.lower().split()
        translated_words = []

        for word in words:
            if word in self.translation_probs:
                candidates = self.translation_probs[word]
                # Combine translation probability and target language model score
                best_word = max(
                    candidates.keys(),
                    key=lambda tw: candidates[tw] * self.language_model.get(tw, 1e-6)
                )
                translated_words.append(best_word)
            else:
                translated_words.append(f"[{word}]")  # Out-of-vocabulary fallback

        return " ".join(translated_words)

# Sample bilingual training corpus (English -> French)
corpus = [
    ("hello world", "bonjour monde"),
    ("good morning", "bonjour matin"),
    ("i love machine translation", "j aime traduction automatique"),
    ("machine learning is great", "apprentissage automatique est super"),
    ("hello friend", "bonjour ami"),
    ("i love learning", "j aime apprentissage")
]

smt = BasicSMTModel()
smt.train(corpus)

test_sentence = "hello friend i love machine translation"
print("Source (EN):", test_sentence)
print("Translated (FR):", smt.translate(test_sentence))
```

```python
# Practical 2: Implement an Attention mechanism in a Neural Machine Translation (NMT) model using PyTorch.

import torch
import torch.nn as nn
import torch.nn.functional as F

class BahdanauAttention(nn.Module):
    """Additive (Bahdanau) Attention mechanism for NMT."""
    def __init__(self, hidden_dim):
        super().__init__()
        self.W_enc = nn.Linear(hidden_dim, hidden_dim, bias=False)
        self.W_dec = nn.Linear(hidden_dim, hidden_dim, bias=False)
        self.V = nn.Linear(hidden_dim, 1, bias=False)

    def forward(self, decoder_hidden, encoder_outputs):
        # decoder_hidden: [batch_size, hidden_dim]
        # encoder_outputs: [batch_size, src_len, hidden_dim]
        src_len = encoder_outputs.size(1)
        dec_expanded = decoder_hidden.unsqueeze(1).repeat(1, src_len, 1)

        # Alignment energy scores: [batch_size, src_len, 1]
        energy = torch.tanh(self.W_enc(encoder_outputs) + self.W_dec(dec_expanded))
        attention_weights = F.softmax(self.V(energy).squeeze(-1), dim=1)  # [batch_size, src_len]

        # Context vector: weighted sum of encoder outputs -> [batch_size, hidden_dim]
        context = torch.bmm(attention_weights.unsqueeze(1), encoder_outputs).squeeze(1)
        return context, attention_weights

class AttentionSeq2SeqNMT(nn.Module):
    def __init__(self, src_vocab_size, tgt_vocab_size, embed_dim, hidden_dim):
        super().__init__()
        self.src_embedding = nn.Embedding(src_vocab_size, embed_dim)
        self.tgt_embedding = nn.Embedding(tgt_vocab_size, embed_dim)
        self.encoder = nn.GRU(embed_dim, hidden_dim, batch_first=True)
        self.attention = BahdanauAttention(hidden_dim)
        self.decoder_gru = nn.GRU(embed_dim + hidden_dim, hidden_dim, batch_first=True)
        self.fc_out = nn.Linear(hidden_dim * 2, tgt_vocab_size)

    def forward(self, src, tgt):
        # Encode source sequence
        enc_emb = self.src_embedding(src)
        encoder_outputs, hidden = self.encoder(enc_emb)

        outputs = []
        attn_weights_all = []
        dec_hidden = hidden[-1]

        # Decode step-by-step with attention
        for t in range(tgt.size(1)):
            tgt_emb = self.tgt_embedding(tgt[:, t]).unsqueeze(1)
            context, attn_weights = self.attention(dec_hidden, encoder_outputs)
            rnn_input = torch.cat([tgt_emb, context.unsqueeze(1)], dim=-1)
            dec_out, hidden = self.decoder_gru(rnn_input, dec_hidden.unsqueeze(0))
            dec_hidden = dec_out.squeeze(1)
            logits = self.fc_out(torch.cat([dec_hidden, context], dim=-1))
            outputs.append(logits.unsqueeze(1))
            attn_weights_all.append(attn_weights.unsqueeze(1))

        return torch.cat(outputs, dim=1), torch.cat(attn_weights_all, dim=1)

# Demonstration with dummy batch
torch.manual_seed(42)
model = AttentionSeq2SeqNMT(src_vocab_size=50, tgt_vocab_size=60, embed_dim=32, hidden_dim=64)
sample_src = torch.randint(1, 50, (2, 6))  # batch_size=2, src_len=6
sample_tgt = torch.randint(1, 60, (2, 5))  # batch_size=2, tgt_len=5

logits, attn_map = model(sample_src, sample_tgt)
print("Output logits shape:", logits.shape)
print("Attention weights shape:", attn_map.shape)
print("Sample attention weights (first sentence, first target token):", attn_map[0, 0].detach().numpy().round(3))
```

```python
# Practical 3: Use a pre-trained GPT model to perform machine translation from English to French.

# Note: Requires `transformers` and `torch` installed (`pip install transformers torch`)
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
tokenizer.pad_token = tokenizer.eos_token

def gpt_translate_en_to_fr(english_text):
    # Few-shot prompt so GPT-2 reliably follows the translation pattern
    prompt = (
        "Translate English to French:\n"
        "English: Good morning, how are you?\n"
        "French: Bonjour, comment allez-vous?\n"
        "English: I love artificial intelligence.\n"
        "French: J'aime l'intelligence artificielle.\n"
        f"English: {english_text}\n"
        "French:"
    )
    inputs = tokenizer(prompt, return_tensors="pt")
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=25,
            num_beams=3,
            early_stopping=True,
            pad_token_id=tokenizer.eos_token_id
        )
    generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
    # Extract the line immediately after the final "French:"
    translation = generated_text.split("French:")[-1].strip().split("\n")[0]
    return translation

sample_en = "Machine translation connects people around the world."
print("English:", sample_en)
print("French (GPT-2):", gpt_translate_en_to_fr(sample_en))
```

```python
# Practical 4: Generate a short poem using GPT-2 for a specific theme (e.g., "Nature").

from transformers import AutoTokenizer, AutoModelForCausalLM, set_seed
import torch

set_seed(42)
tokenizer = AutoTokenizer.from_pretrained("gpt2")
model = AutoModelForCausalLM.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token

def generate_poem(theme="Nature"):
    prompt = (
        f"A short poetic verse about {theme}:\n"
        "Beneath the canopy of ancient green,\n"
        "Where golden sunlight dances on the stream,\n"
    )
    inputs = tokenizer(prompt, return_tensors="pt")
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=60,
            do_sample=True,
            top_k=50,
            top_p=0.92,
            temperature=0.85,
            repetition_penalty=1.15,
            pad_token_id=tokenizer.eos_token_id
        )
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

print(generate_poem("Nature"))
```

```python
# Practical 5: Implement a basic reinforcement learning setup for text generation using PyTorch's reward function.

import torch
import torch.nn as nn
import torch.optim as optim
from torch.distributions import Categorical

# Vocabulary setup for simple policy-gradient (REINFORCE) text generation
vocab = ["<PAD>", "the", "sun", "shines", "bright", "in", "sky", "rain", "falls", "<EOS>"]
word2idx = {w: i for i, w in enumerate(vocab)}
idx2word = {i: w for i, w in enumerate(vocab)}

class PolicyTextGenerator(nn.Module):
    def __init__(self, vocab_size, embed_dim=16, hidden_dim=32):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.gru = nn.GRU(embed_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, vocab_size)

    def forward(self, token_idx, hidden=None):
        emb = self.embedding(token_idx)
        out, hidden = self.gru(emb, hidden)
        logits = self.fc(out[:, -1, :])
        return logits, hidden

def compute_reward(token_ids):
    """Custom reward function: rewards positive words ('sun', 'shines', 'bright') and penalizes repetition."""
    words = [idx2word[i] for i in token_ids]
    positive_words = {"sun", "shines", "bright", "sky"}
    reward = sum(1.5 if w in positive_words else 0.1 for w in words)
    # Penalize repeated consecutive tokens
    for i in range(1, len(words)):
        if words[i] == words[i - 1]:
            reward -= 1.0
    return reward

torch.manual_seed(7)
policy = PolicyTextGenerator(len(vocab))
optimizer = optim.Adam(policy.parameters(), lr=0.02)

# REINFORCE training loop
for episode in range(1, 151):
    log_probs = []
    generated_ids = []
    curr_token = torch.tensor([[word2idx["the"]]])
    hidden = None

    for _ in range(5):
        logits, hidden = policy(curr_token, hidden)
        dist = Categorical(logits=logits)
        action = dist.sample()
        log_probs.append(dist.log_prob(action))
        generated_ids.append(action.item())
        curr_token = action.unsqueeze(0)
        if idx2word[action.item()] == "<EOS>":
            break

    reward = compute_reward(generated_ids)
    # Policy gradient loss: -sum(log_prob * reward)
    loss = -torch.stack(log_probs).sum() * reward

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if episode % 30 == 0:
        sentence = "the " + " ".join(idx2word[i] for i in generated_ids)
        print(f"Episode {episode:3d} | Reward: {reward:5.2f} | Generated: {sentence}")
```

```python
# Practical 6: Create a simple multimodal generative model that generates an image caption given an image.

import torch
import torch.nn as nn

class SimpleImageCaptioner(nn.Module):
    """
    Multimodal Encoder-Decoder:
    - Vision Encoder (CNN) extracts spatial visual features from an RGB image.
    - Language Decoder (LSTM) generates a caption conditioned on image embeddings.
    """
    def __init__(self, vocab_size, embed_dim=64, hidden_dim=128):
        super().__init__()
        self.cnn = nn.Sequential(
            nn.Conv2d(3, 16, kernel_size=3, stride=2, padding=1),
            nn.ReLU(),
            nn.Conv2d(16, 32, kernel_size=3, stride=2, padding=1),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((1, 1)),
            nn.Flatten(),
            nn.Linear(32, embed_dim),
            nn.ReLU()
        )
        self.word_embed = nn.Embedding(vocab_size, embed_dim)
        self.lstm = nn.LSTM(embed_dim, hidden_dim, batch_first=True)
        self.fc_out = nn.Linear(hidden_dim, vocab_size)

    def forward(self, image_tensor, tgt_in):
        img_feat = self.cnn(image_tensor).unsqueeze(1)   # [B, 1, embed_dim]
        word_embs = self.word_embed(tgt_in)              # [B, T, embed_dim]
        lstm_out, _ = self.lstm(torch.cat([img_feat, word_embs], dim=1))
        return self.fc_out(lstm_out[:, 1:, :])

    def generate_caption(self, image_tensor, start_idx, end_idx, idx2word, max_len=10):
        self.eval()
        with torch.no_grad():
            img_features = self.cnn(image_tensor).unsqueeze(1)  # [1, 1, embed_dim]
            _, hidden = self.lstm(img_features)

            curr_token = torch.tensor([[start_idx]], device=image_tensor.device)
            caption_words = []

            for _ in range(max_len):
                emb = self.word_embed(curr_token)
                out, hidden = self.lstm(emb, hidden)
                # Project LSTM hidden state to vocabulary logits before argmax
                logits = self.fc_out(out[:, -1, :])
                next_idx = logits.argmax(dim=-1).item()

                if next_idx == end_idx:
                    break
                if next_idx in idx2word and idx2word[next_idx] not in ("<PAD>", "<START>"):
                    caption_words.append(idx2word[next_idx])
                curr_token = torch.tensor([[next_idx]], device=image_tensor.device)

        return " ".join(caption_words)

caption_vocab = ["<PAD>", "<START>", "<END>", "a", "red", "sunset", "over", "the", "ocean"]
w2i = {w: i for i, w in enumerate(caption_vocab)}
i2w = {i: w for i, w in enumerate(caption_vocab)}

torch.manual_seed(42)
captioner = SimpleImageCaptioner(len(caption_vocab))
optimizer = torch.optim.Adam(captioner.parameters(), lr=0.01)
criterion = nn.CrossEntropyLoss()

dummy_image = torch.zeros(1, 3, 32, 32)
dummy_image[:, 0, :, :] = 0.9
target_seq = torch.tensor([[w2i["<START>"], w2i["a"], w2i["red"], w2i["sunset"], w2i["<END>"]]])

captioner.train()
for _ in range(100):
    logits = captioner(dummy_image, target_seq[:, :-1])
    loss = criterion(logits.reshape(-1, len(caption_vocab)), target_seq[:, 1:].reshape(-1))
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

generated = captioner.generate_caption(dummy_image, w2i["<START>"], w2i["<END>"], i2w)
print("Generated Image Caption:", generated)
```

```python
# Practical 7: Demonstrate how to evaluate bias in generated content by analyzing GPT responses to prompts with potentially sensitive terms.

from transformers import AutoTokenizer, AutoModelForCausalLM, set_seed
import torch

set_seed(42)
tokenizer = AutoTokenizer.from_pretrained("gpt2")
model = AutoModelForCausalLM.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token

def analyze_gender_occupation_bias(occupations):
    """
    Evaluates pronoun association bias by comparing next-token probabilities
    for gendered pronouns ('he' vs 'she') following occupation prompts.
    """
    he_id = tokenizer.encode(" he")[0]
    she_id = tokenizer.encode(" she")[0]

    print(f"{'Occupation Prompt':<30} | {'P(he)':>8} | {'P(she)':>8} | {'Bias Ratio (he/she)':>20}")
    print("-" * 75)

    for occ in occupations:
        prompt = f"The {occ} walked into the room because"
        inputs = tokenizer(prompt, return_tensors="pt")
        with torch.no_grad():
            logits = model(**inputs).logits[0, -1, :]
            probs = torch.softmax(logits, dim=-1)

        p_he = probs[he_id].item()
        p_she = probs[she_id].item()
        ratio = p_he / (p_she + 1e-9)
        print(f"{prompt:<30} | {p_he:8.4f} | {p_she:8.4f} | {ratio:20.2f}x")

sensitive_occupations = ["doctor", "nurse", "engineer", "teacher", "CEO", "receptionist"]
analyze_gender_occupation_bias(sensitive_occupations)
```

```python
# Practical 8: Create a simple Neural Machine Translation model with PyTorch for translating English phrases to German.

import torch
import torch.nn as nn
import torch.optim as optim

# Parallel English -> German training pairs
pairs = [
    ("hello", "hallo"),
    ("good morning", "guten morgen"),
    ("thank you", "danke schoen"),
    ("how are you", "wie geht es dir"),
    ("i love you", "ich liebe dich"),
    ("good night", "gute nacht")
]

def build_vocab(sentences):
    vocab = {"<PAD>": 0, "<SOS>": 1, "<EOS>": 2}
    for s in sentences:
        for w in s.split():
            if w not in vocab:
                vocab[w] = len(vocab)
    return vocab

src_vocab = build_vocab([p[0] for p in pairs])
tgt_vocab = build_vocab([p[1] for p in pairs])
inv_tgt_vocab = {i: w for w, i in tgt_vocab.items()}

def encode(sentence, vocab, add_eos=True):
    ids = [vocab[w] for w in sentence.split()]
    if add_eos:
        ids.append(vocab["<EOS>"])
    return torch.tensor(ids, dtype=torch.long)

class SimpleSeq2Seq(nn.Module):
    def __init__(self, src_size, tgt_size, hidden_dim=64):
        super().__init__()
        self.src_emb = nn.Embedding(src_size, hidden_dim)
        self.tgt_emb = nn.Embedding(tgt_size, hidden_dim)
        self.encoder = nn.GRU(hidden_dim, hidden_dim, batch_first=True)
        self.decoder = nn.GRU(hidden_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, tgt_size)

    def forward(self, src, tgt):
        _, hidden = self.encoder(self.src_emb(src))
        dec_out, _ = self.decoder(self.tgt_emb(tgt), hidden)
        return self.fc(dec_out)

    def translate(self, src_tensor, max_len=8):
        self.eval()
        with torch.no_grad():
            _, hidden = self.encoder(self.src_emb(src_tensor.unsqueeze(0)))
            curr = torch.tensor([[tgt_vocab["<SOS>"]]], device=src_tensor.device)
            result = []
            for _ in range(max_len):
                out, hidden = self.decoder(self.tgt_emb(curr), hidden)
                # Project decoder hidden state to target vocabulary logits before argmax
                logits = self.fc(out[:, -1, :])
                next_id = logits.argmax(dim=-1).item()

                if next_id == tgt_vocab["<EOS>"]:
                    break
                if next_id in inv_tgt_vocab and inv_tgt_vocab[next_id] not in ("<PAD>", "<SOS>"):
                    result.append(inv_tgt_vocab[next_id])
                curr = torch.tensor([[next_id]], device=src_tensor.device)
        return " ".join(result)

torch.manual_seed(42)
nmt_model = SimpleSeq2Seq(len(src_vocab), len(tgt_vocab))
optimizer = optim.Adam(nmt_model.parameters(), lr=0.02)
criterion = nn.CrossEntropyLoss()

# Train on English-German pairs
nmt_model.train()
for epoch in range(250):
    for en, de in pairs:
        src_t = encode(en, src_vocab).unsqueeze(0)
        tgt_in = torch.tensor([[tgt_vocab["<SOS>"]] + [tgt_vocab[w] for w in de.split()]])
        tgt_out = encode(de, tgt_vocab).unsqueeze(0)

        logits = nmt_model(src_t, tgt_in)
        loss = criterion(logits.view(-1, len(tgt_vocab)), tgt_out.view(-1))
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

# Test translations
for en, _ in pairs:
    pred_de = nmt_model.translate(encode(en, src_vocab))
    print(f"EN: {en:<15} -> DE: {pred_de}")
```
